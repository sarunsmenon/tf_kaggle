```python
import pandas as pd
import os
import json
import collections
import shutil
from PIL import Image
```

```python
base_folder_nm = "C:/Nuras/Kaggle/Benetech Making Graphs/benetech-making-graphs-accessible/"
bar_chart_folder = base_folder_nm+ "/bar_Chart/"
dot_chart_folder = base_folder_nm+ "/dot_Chart/"
line_chart_folder = base_folder_nm+ "/line_Chart/"
scatter_chart_folder = base_folder_nm+ "/scatter_Chart/"
img_fldr_lst = [bar_chart_folder, dot_chart_folder, line_chart_folder, scatter_chart_folder]
```

```python
size_df_lst = []

for fldr in img_fldr_lst:
    for fl in os.listdir(fldr):
        w,h = Image.open(fldr+fl).size
        df = pd.DataFrame({'width':[w], 'height' : [h]})
        size_df_lst.append(df)
```

```python
size_df = pd.concat(size_df_lst)
```

```python
print(min(size_df['width']))
print(min(size_df['height']))
```
