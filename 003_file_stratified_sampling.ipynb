```python
import json
import pandas as pd
import numpy as np
import os
import shutil
```

```python
base_path = "C:/Nuras/Kaggle/Benetech Making Graphs/benetech-making-graphs-accessible/"
annotation_path =base_path + "train/annotations/"
```

```python
chart_dtl_lst =[]
for file_nm in os.listdir(annotation_path):
    file_data_dict = json.load(open(annotation_path+file_nm, "r"))
    chart_type = file_data_dict['chart-type']
    chart_name = file_nm.split(".")[0]
    chart_df = pd.DataFrame({'chart_type' : [chart_type], 'chart_name':[chart_name]} )
    chart_dtl_lst.append(chart_df)
```

```python
chart_dtl_df = pd.concat(chart_dtl_lst)
```

```python
chart_dtl_sample_df = chart_dtl_df.groupby('chart_type',group_keys=False).apply(lambda x:x.sample(frac= 0.1))
```

```python
def sample_copy(chart_type , chart_name):
    if 'bar' in chart_type:
        prefix="bar"
    elif 'scatter' in chart_type:
        prefix="scatter"
    elif 'dot' in chart_type:
        prefix="dot"
    else:
        prefix="line"

    src_file = base_path + prefix+"_chart/"+chart_name+".jpg"
    dest_fldr = base_path + prefix+ "_chart_sample/"
    shutil.copy(src_file, dest_fldr)

```

```python
chart_dtl_sample_df.apply(lambda x : sample_copy(x['chart_type'], x['chart_name']), axis=1)
```
