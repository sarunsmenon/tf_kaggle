```python
import pandas as pd
import os
import json
import collections
import shutil
```

```python
base_folder_nm = "C:/Nuras/Kaggle/Benetech Making Graphs/benetech-making-graphs-accessible/"
train_images_path = base_folder_nm + "train/images/"
annotatations_path = base_folder_nm + "train/annotations/"
```

```python
os.mkdir(base_folder_nm+"bar_chart")
os.mkdir(base_folder_nm+"line_chart")
os.mkdir(base_folder_nm+"scatter_chart")
os.mkdir(base_folder_nm+"dot_chart")
```

```python
for annot_file in os.listdir(annotatations_path):
    with open(annotatations_path+annot_file,"r") as f:
        file_content = f.read()

    file_cont_dict = json.loads(file_content)
    img_nm = annot_file.replace(".json",".jpg")

    if "bar" in file_cont_dict['chart-type'] :
        shutil.move(train_images_path+img_nm, base_folder_nm+"bar_chart")
    elif "line" in file_cont_dict['chart-type'] :
        shutil.move(train_images_path+img_nm, base_folder_nm+"line_chart")
    elif "scatter" in file_cont_dict['chart-type'] :
        shutil.move(train_images_path+img_nm, base_folder_nm+"scatter_chart")
    elif"dot" in file_cont_dict['chart-type'] :
        shutil.move(train_images_path+img_nm, base_folder_nm+"dot_chart")
```
